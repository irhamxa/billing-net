```python

import numpy as np

# Membuat tiga buah vektor
A = np.array([1, 2, -1])
B = np.array([4, 2, 3])
C = np.array([0, 2, -3])

# a. A . (B x C)
hasil_a = np.dot(A, np.cross(B, C))

# b. A . (B + C)
hasil_b = np.dot(A, B + C)

# c. A x (B + C)
hasil_c = np.cross(A, B + C)

# Menampilkan hasil
print("a. A . (B x C) =", hasil_a)
print("b. A . (B + C) =", hasil_b)
print("c. A x (B + C) =", hasil_c)
```

```python

import numpy as np

A = np.array([
    [3, 0],
    [-1, 2],
    [1, 1]
])

B = np.array([
    [4, -1],
    [0, 2]
])

C = np.array([
    [1, 4, 2],
    [3, 1, 5]
])

D = np.array([
    [1, 5, 2],
    [-1, 0, 1],
    [3, 2, 4]
])

E = np.array([
    [6, 1, 3],
    [-1, 1, 2],
    [4, 1, 3]
])

# Operasi matriks
print("1. D + E =")
print(D + E)

print("\n2. D - E =")
print(D - E)

print("\n3. 5A =")
print(5 * A)

print("\n4. (AB)C =")
print((A @ B) @ C)

print("\n5. (DA)^T =")
print((D @ A).T)

```

```python

# Tabel menggunakan Python Murni

judul = "Tabel 4.1 Geometri Peledakan pada Masing-masing Blok"

header = [
    "Lokasi", "Jumlah Lubang", "S (m)", "B (m)",
    "H (m)", "T (m)", "D (mm)", "PC (m)"
]

data = [
    ["OP - 16", "41", "3,5", "2,5", "5,7", "1,9", "89", "3,8"],
    ["HI - 15/16", "10", "4,5", "3", "9,1", "3,5", "114", "5,6"],
    ["O - 16", "36", "3,5", "2,5", "5,7", "1,9", "89", "3,8"],
    ["P - 16", "33", "3,5", "2,5", "6,0", "2,0", "89", "4,0"],
    ["O - 16", "30", "3,5", "2,5", "5,5", "2,0", "89", "3,5"],
    ["L - 14", "25", "4,5", "3,5", "6,0", "2,5", "114", "3,5"],
    ["M - 13", "25", "4,5", "3,5", "6,0", "2,5", "114", "3,5"],
    ["T - 16", "31", "3,5", "2,5", "5,5", "1,9", "89", "3,6"]
]

# Menggabungkan header dan data
semua_data = [header] + data

# Menghitung lebar setiap kolom
lebar = [
    max(len(str(baris[i])) for baris in semua_data)
    for i in range(len(header))
]

# Menampilkan judul
print(judul.center(sum(lebar) + 3 * len(header)))

# Membuat garis tabel
garis = "+".join("-" * (w + 2) for w in lebar)
print("+" + garis + "+")

# Menampilkan header
print("|" + "|".join(
    f" {header[i]:<{lebar[i]}} "
    for i in range(len(header))
) + "|")

print("+" + garis + "+")

# Menampilkan isi tabel
for baris in data:
    print("|" + "|".join(
        f" {baris[i]:<{lebar[i]}} "
        for i in range(len(header))
    ) + "|")

print("+" + garis + "+")
```
